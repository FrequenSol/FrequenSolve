## Interpret the result
The pressure receiver values are complex frequency-domain amplitudes at 10 Hz, not a time-domain seismogram. Compare amplitude and phase along receiver position; inspect the VTK pressure field and layer interface.
Retain the run ID and authored inputs for reproduction. A converged solve does not by itself validate your scientific model.

```python
traces = result.traces()
print(traces.summary)
group = traces.groups[0]
component = traces.components(group)[0]
source = traces.sources(group)[0]
response = traces.fd(group, component, source)
abs(response).plot()
```
